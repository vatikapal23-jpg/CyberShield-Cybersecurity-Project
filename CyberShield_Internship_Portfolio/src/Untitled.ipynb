```python
import os

matches = []

for root, dirs, files in os.walk(os.getcwd()):
    dirs[:] = [d for d in dirs if d not in {".venv", "__pycache__"}]
    if "app.py" in files and "requirements.txt" in files:
        matches.append(root)

if matches:
    print("Project folder found:")
    for folder in matches:
        print(folder)
else:
    print("Project folder not found.")
    print("Please extract the CyberShield ZIP file first.")
```

```python
import os

search_locations = [
    os.getcwd(),
    os.path.expanduser("~/Downloads"),
    os.path.expanduser("~/Desktop"),
    os.path.expanduser("~")
]

for location in search_locations:
    if os.path.isdir(location):
        print(f"\nChecking: {location}")
        for item in os.listdir(location):
            if "CyberShield" in item:
                print("  Found:", os.path.join(location, item))
```

```python
import zipfile
import os

zip_path = r"ZIP_PATH"
extract_to = os.path.join(os.getcwd(), "CyberShield_Project")

with zipfile.ZipFile(zip_path, "r") as z:
    z.extractall(extract_to)

print("Extracted to:", extract_to)
print("Contents:", os.listdir(extract_to))
```
